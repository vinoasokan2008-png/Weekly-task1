```python
# Number of rows before removing null values
print("Rows before removing null values:", df.shape[0])

# Remove rows containing null values
df = df.dropna()

# Number of rows after removing null values
print("Rows after removing null values:", df.shape[0])
```

```python
print("Number of duplicate rows:",df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Dataset shape after removing duplicates:",df.shape)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df[["Order Date", "Ship Date"]].dtypes
```

```python
text_columns = ["Category", "Sub-Category", "Segment"]

for column in text_columns:
    df[column] = (
        df[column]
       .astype(str)
       .str.strip()
       .str.title()
    )
```

```python
for column in text_columns:
    print("\n",column)
    print(df[column].unique())
```

```python
print("Final Shape:", df.shape)
print("\nTotal Missing Values:", df.isnull().sum().sum())
print("Total Duplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
numerical_features = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

df[numerical_features].describe()
```

```python
summary = pd.DataFrame({
    "Mean": df[numerical_features].mean(),
    "Median": df[numerical_features].median(),
    "Minimum": df[numerical_features].min(),
    "Maximum": df[numerical_features].max(),
    "Standard Deviation": df[numerical_features].std()
})

summary
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
   .sum()
   .sort_values(ascending=False)
)

category_sales
```

```python
category_profit = (
    df.groupby("Category")["Profit"]
   .sum()
   .sort_values(ascending=False)
)

category_profit
```

```python
sub_category_profit = (
    df.groupby("Sub-Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
sub_category_profit
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Category",
    y="Sales",
    estimator="sum"
)

plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Category",
    y="Profit",
    estimator="sum"
)

plt.title("Category-wise Total Profit")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    y="Sales"
)
plt.title("Sales Distribution")
plt.ylabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    y="Profit"
)
plt.title("Profit Distribution")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Category",
    y="Profit"
)

plt.title("Profit Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(8, 5))


sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit"
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")



plt.show()
```

```python
df["Profit Margin"] = (df["Profit"]/df["Sales"])*100
df[["Sales", "Profit","Profit Margin"]].head()
```

```python
plt.figure(figsize=(8, 5))


sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit Margin"
)

plt.title("Discount vs Profit Margin")
plt.xlabel("Discount")
plt.ylabel("Profit Margin (%)")



plt.show()
```

```python
numerical_df = df.select_dtypes(include=np.number)

correlation_matrix = numerical_df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Numerical Attributes")
plt.show()
```
