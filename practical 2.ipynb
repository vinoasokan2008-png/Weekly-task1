```python
##### PRACTICAL 2
# Conditional Statements and Loops

# 1. Check Positive Number (if)
num = int(input("Enter a number: "))
if num > 0:
    print("Positive Number")

# 2. Even or Odd (if-else)
num = int(input("Enter another number: "))
if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")

# 3. Grade (if-elif-else)
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
else:
    print("Fail")

# 4. Login (Nested if-else)
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "1234":
        print("Login Successful")
    else:
        print("Incorrect Password")
else:
    print("Invalid Username")

# 5. Voting Eligibility (One-line if-else)
age = int(input("Enter your age: "))
print("Eligible to Vote" if age >= 18 else "Not Eligible to Vote")

# 6. For Loop
print("Numbers using for loop:")
for i in range(1, 6):
    print(i)

# 7. While Loop
print("Numbers using while loop:")
i = 1
while i <= 5:
    print(i)
    i += 1

# 8. Break Statement
print("Break Example:")
for i in range(1, 11):
    if i == 5:
        break
    print(i)

# 9. Continue Statement
print("Continue Example:")
for i in range(1, 11):
    if i == 5:
        continue
    print(i)

# 10. Pass Statement
print("Pass Example:")
for i in range(1, 6):
    if i == 3:
        pass
    print(i)

# 11. Match-Case Statement
day = int(input("Enter day number (1-7): "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid Day")

print("Program Ended")
```

```python
# Practical 3 - User Defined Functions
# 1. No arguments, No return value
def greet():
    print("Hello!")
greet()
# 2. Arguments, No return value
def add(a, b):
    print("Sum =", a + b)

add(10, 20)

# 3. No arguments, Return value
def message():
    return "Welcome to Python"

print(message())

# 4. Arguments, Return value
def multiply(a, b):
    return a * b

print("Product =", multiply(5, 4))

# 5. Default argument
def student(name="Vino"):
    print("Name:", name)

student()
student("Ravi")

# 6. Keyword arguments
def details(name, age):
    print("Name:", name)
    print("Age:", age)

details(age=18, name="Anu")

# 7. Variable length arguments
def total(*marks):
    print("Total Marks =", sum(marks))

total(80, 90, 85)

# 8. Lambda function
square = lambda x: x * x
print("Square =", square(5))

# 9. Recursive function
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print("Factorial =", fact(5))

# 10. Nested function
def outer():
    def inner():
        print("Inner Function")
    inner()

outer()
```

```python
# Create a list
nums = [1, 2, 3, 4, 5]

print("Original List:", nums)
print("First Element:", nums[0])
print("Last Element:", nums[-1])

# Append
nums.append(6)
print("After Append:", nums)

# Insert
nums.insert(2, 10)
print("After Insert:", nums)

# Remove
nums.remove(10)
print("After Remove:", nums)

# Length
print("Length:", len(nums))

# Traverse
print("List Elements:")
for i in nums:
    print(i)

# List Comprehension
square = [x*x for x in nums]
even = [x for x in nums if x % 2 == 0]
odd = [x for x in nums if x % 2 != 0]
double = [x*2 for x in nums]

matrix = [[1, 2], [3, 4]]
flat = [x for row in matrix for x in row]

print("Squares:", square)
print("Even:", even)
print("Odd:", odd)
print("Double:", double)
print("Flattened List:", flat)
```

```python
# Indexing, Slicing and Reshaping in NumPy

import numpy as np

# Creating NumPy Array
array1 = np.array([10, 20, 30, 40, 50, 60])

print("Original Array:")
print(array1)

# ---------------- INDEXING ----------------

print("\nIndexing Operations")

print("First Element:", array1[0])
print("Third Element:", array1[2])
print("Last Element:", array1[-1])

# ---------------- SLICING ----------------

print("\nSlicing Operations")

print("Elements from Index 1 to 4:")
print(array1[1:5])

print("Elements from Beginning to Index 3:")
print(array1[:4])

print("Elements from Index 2 to End:")
print(array1[2:])

print("Elements with Step Size 2:")
print(array1[::2])

# ---------------- 2D ARRAY ----------------

array2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("\n2D Array:")
print(array2)

# 2D Array Indexing
print("\nElement at Row 1 Column 2:")
print(array2[0, 1])

print("Element at Row 2 Column 3:")
print(array2[1, 2])

# 2D Array Slicing
print("\nFirst Row:")
print(array2[0])

print("Second Column:")
print(array2[:, 1])

# ---------------- RESHAPING ----------------

array3 = np.array([1, 2, 3, 4, 5, 6])

print("\nOriginal 1D Array:")
print(array3)

# Reshape into 2 Rows and 3 Columns
reshaped1 = array3.reshape(2, 3)

print("\nReshaped Array (2x3):")
print(reshaped1)

# Reshape into 3 Rows and 2 Columns
reshaped2 = array3.reshape(3, 2)

print("\nReshaped Array (3x2):")
print(reshaped2)

# Flattening Array
flattened = reshaped1.flatten()

print("\nFlattened Array:")
print(flattened)

# Array Attributes
print("\nShape of Array:")
print(reshaped1.shape)

print("\nDimensions of Array:")
print(reshaped1.ndim)

print("\nSize of Array:")
print(reshaped1.size)
```

```python
# practical 14

# Merge and Join Dataframes using Pandas

import pandas as pd

# ---------------- CONCATENATE OPERATIONS ----------------
# (All columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])

print("\nDataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])

print("\nDataFrame 2:")
print(df2)

# Concatenate DataFrames
frames = [df1, df2]
result = pd.concat(frames)

print("Concatenate DataFrames:")
print(result)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan"]
}, index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)


```
